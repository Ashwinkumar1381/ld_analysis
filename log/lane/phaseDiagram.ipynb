```python
Fd_list = [30, 50, 70, 80, 90, 100, 150, 200, 250, 300, 350, 400, 450, 500]
tau_list = ["5e-2"]

fig, ax = plt.subplots(figsize = (5, 5))
ms = 8
lw = 1.2

for i in range(len(tau_list)):
    tau_list_val = liteval(tau_list[i])

    Phi_list = []
    for j in range(len(Fd_list)):
        fpath = os.path.join(dirpath, "Fd{val}".format(val = Fd_list[j]), "avgOrder.dat")
        [data, _] = readData(fpath, 1)

        tau = data[1]
        order = data[4]

        if(tau_list_val in tau):
            Index = tau.index(tau_list_val)
            Phi_list.append(order[Index])

    print(len(Fd_list), len(Phi_list))
    print(Phi_list)

    ax.plot(Fd_list, Phi_list, "o-", color = "crimson", mec = "k", mew = 0.9, ms = ms, lw = lw, 
            label = r"${val}$".format(val = float2latex([tau_list[i]])))

ax.set(xlim = (0, 510), ylim = (0, 1.05))
ax.set_xlabel(r"$Pe_d$", fontsize = 21)
ax.set_ylabel(r"$\langle\phi_L\rangle$", fontsize = 21)
ax.tick_params(axis = "both", which = "major", direction = "in", labelsize = 15)
leg = ax.legend(title = r"$\tau_{in}^*$", frameon = False, fontsize = 15)
leg.get_title().set_fontsize("18")

# fig.savefig(os.path.join(figpath, "reentrant.png"), dpi = 500, bbox_inches = "tight")
```
