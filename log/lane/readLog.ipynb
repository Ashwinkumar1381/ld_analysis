```python
# Extracts columns of thermo data from a single log.lammps and writes to thermo.dat

run_nr = [2]
interval = int(10)

fname1 = "/home/ashwin/Desktop/ashwin_md/LD/lmp/tests/sem_analysis/log.lammps"
fobj1 = open(fname1, mode = "r", encoding = "utf-8")

fname2 = fname1.removesuffix("/log.lammps") + "/thermo.dat"
fobj2 = open(fname2, mode = "w")

ctr = 0
run_ctr = 0 
key = 0

fobj2.write("Step PotEng KinEng TotEng Temp")

for line in fobj1:
    array = line2array(line)

    if(array[0] == 'Loop'):
        if(key == 1):
            key = 0
            if(ctr == len(run_nr)):
                break

    if(key == 1):
        if(liteval(array[0])%interval == 0):
            line = array2line(array)
            fobj2.write("\n" + line)

    if(array[0] == 'Step'):
        run_ctr += 1
        print(run_ctr, run_nr[ctr])
        if(run_ctr == run_nr[ctr]):
            key = 1
            ctr += 1

print("\nThermo data extracted successfully.")

fobj1.close()
fobj2.close()
```

```python
# Extracts columns of thermo data from 2 LAMMPS log files (log2.lammps and log.lammps) and writes to thermo.dat

run_nr = [1, 2, 3, 4, 5]
break_step = 9440000000
interval = int(1e5)

fname1 = "//media/ashwin/One Touch/ashwin_md/lane/May2025/lmp/Data30/log2.lammps"
fobj1 = open(fname1, mode = "r", encoding = "utf-8")

fname2 = fname1.removesuffix("log2.lammps") + "thermo.dat"
fobj2 = open(fname2, mode = "w")

ctr = 0
run_ctr = 0 
key = 0

fobj2.write("Step Temp PotEng KinEng TotEng\n")

for line in fobj1:
    array = line2array(line)

    if(array[0] == 'Loop'):
        if(key == 1):
            key = 0
            if(ctr == len(run_nr) - 1):
                break

    if(key == 1):
        if(liteval(array[0])%interval == 0):
            line = array2line(array)
            fobj2.write(line + "\n")
        if(liteval(array[0]) == break_step):
            break

    if(array[0] == 'Step'):
        run_ctr += 1
        print(run_ctr, run_nr[ctr])
        if(run_ctr == run_nr[ctr]):
            key = 1
            ctr += 1

fobj1.close()

fname1 = fname1.removesuffix("log2.lammps") + "log.lammps"
fobj1 = open(fname1, mode = "r", encoding = "utf-8")

run_nr = [1]
ctr = 0
run_ctr = 0 
key = 0

for line in fobj1:
    array = line2array(line)

    if(array[0] == 'Loop'):
        if(key == 1):
            key = 0

    if(key == 1):
        array[0] = str(liteval(array[0]) + break_step)
        line = array2line(array)
        fobj2.write(line + "\n")

    if(array[0] == 'Step'):
        run_ctr += 1
        if(run_ctr == run_nr[ctr]):
            key = 1
            ctr += 1

fobj2.close()
print("Thermo data extracted successfully.\n")
```
