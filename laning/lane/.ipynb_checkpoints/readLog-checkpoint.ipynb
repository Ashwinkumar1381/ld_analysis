```python
import matplotlib.pyplot as plt
plt.rc('text', usetex = True)

from ast import literal_eval as liteval
import numpy as np
import os

import sys
sys.path.insert(0, os.getcwd().removesuffix("/lane"))
from file_utils import *
sys.path.pop(0)

dirpath = "//media/ashwin/Expansion/ashwin_md/lane/Aug_Sept2025/"
figpath = "/home/ashwin/Desktop/ashwin_md/LD/imgs/June2025/"
```

```python
# Extracts columns of thermo data from a single log.lammps file and writes to thermo.dat

run_nr = [1, 2, 3, 4, 5]
interval = int(1e5)
Fd = "Fd100"
tau = "tau_1e0"

fname1 = os.path.join(dirpath, Fd, tau, "log.lammps")
fobj1 = open(fname1, mode = "r", encoding = "utf-8")

fname2 = os.path.join(dirpath, Fd, tau, "thermo.dat")
fobj2 = open(fname2, mode = "w")

ctr = 0
run_ctr = -1
key = 0

fobj2.write("Step Temp PotEng KinEng TotEng")

curr_step = 0
for line in fobj1:
    array = line2array(line)

    if(array[0] == 'Loop'):
        if(key == 1):
            key = 0
            if(ctr == len(run_nr)):
                break

    if(key == 1):
        if(liteval(array[0]) == curr_step):
            continue
        elif(liteval(array[0])%interval == 0):
            line = array2line(array)
            fobj2.write("\n" + line)
            
        curr_step = liteval(array[0])

    if(array[0] == 'Step'):
        run_ctr += 1
        print(run_ctr, run_nr[ctr])
        if(run_ctr == run_nr[ctr]):
            key = 1
            ctr += 1

print("\nThermo data extracted successfully.")

fobj1.close()
fobj2.close()
```

```python
# Extracts columns of thermo data from 2 LAMMPS log files (log_old.lammps and log.lammps) and writes to thermo.dat

run_nr = [1, 2, 3, 4, 5]
break_step = 4710000000
interval = int(1e5)
Fd = "Fd100"
tau = "tau_1e-3"

fname1 = os.path.join(dirpath, Fd, tau, "log_old.lammps")
fobj1 = open(fname1, mode = "r", encoding = "utf-8")

fname2 = os.path.join(dirpath, Fd, tau, "thermo.dat")
fobj2 = open(fname2, mode = "w")

ctr = 0
run_ctr = -1 
key = 0

fobj2.write("Step Temp PotEng KinEng TotEng\n")

curr_step = 0
for line in fobj1:
    array = line2array(line)

    if(array[0] == 'Loop'):
        if(key == 1):
            key = 0
            if(ctr == len(run_nr) + 1):
                break

    if(key == 1):
        if(liteval(array[0]) == curr_step):
            continue
        elif(liteval(array[0]) == break_step):
            break
        elif(liteval(array[0])%interval == 0):
            line = array2line(array)
            fobj2.write("\n" + line)
            
        curr_step = liteval(array[0])
        
    if(array[0] == 'Step'):
        run_ctr += 1
        print(run_ctr, run_nr[ctr])
        if(run_ctr == run_nr[ctr]):
            key = 1
            ctr += 1

fobj1.close()

fname1 = os.path.join(dirpath, Fd, tau, "log.lammps")
fobj1 = open(fname1, mode = "r", encoding = "utf-8")

run_nr = [1]
ctr = 0
run_ctr = 0 
key = 0

for line in fobj1:
    array = line2array(line)

    if(array[0] == 'Loop'):
        if(key == 1):
            key = 0

    if(key == 1):
        array[0] = str(liteval(array[0]) + break_step)
        line = array2line(array)
        fobj2.write(line + "\n")

    if(array[0] == 'Step'):
        run_ctr += 1
        if(run_ctr == run_nr[ctr]):
            key = 1
            ctr += 1

fobj2.close()
print("Thermo data extracted successfully.\n")
```
